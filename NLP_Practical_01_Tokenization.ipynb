```python
import nltk
import spacy

# Download required NLTK resources
nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.tokenize import sent_tokenize, word_tokenize
```

```python
# Download English spaCy model
!python -m spacy download en_core_web_sm
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
text = """
Natural Language Processing is a branch of Artificial Intelligence.
It helps computers understand human language.
NLT​​K and spaCy are popular NLP libraries.
"""

print(text)
```

```python
nltk_sentences = sent_tokenize(text)

print("Sentences using NLTK:")
for i, sentence in enumerate(nltk_sentences, 1):
    print(f"{i}. {sentence}")
```

```python
nltk_words = word_tokenize(text)

print("Words using NLTK:")
print(nltk_words)
```

```python
doc = nlp(text)

print("Sentences using spaCy:")
for i, sentence in enumerate(doc.sents, 1):
    print(f"{i}. {sentence.text}")
```

```python
spacy_words = [token.text for token in doc]

print("Words using spaCy:")
print(spacy_words)
```

```python
print("Comparison")
print("-" * 40)

print(f"NLTK sentence count  : {len(nltk_sentences)}")
print(f"spaCy sentence count : {len(list(doc.sents))}")

print(f"\nNLTK word count      : {len(nltk_words)}")
print(f"spaCy word count     : {len(spacy_words)}")
```

```python
## Result

#The given text was successfully tokenized into sentences and words using both NLTK and spaCy.

#NLTK and spaCy were used to demonstrate sentence-level and word-level tokenization.
```
